### Exercice 1 – Racines complexes
Résoudre dans $\mathbb{C}$ l’équation $z^3 = -8$.  
Donner les solutions sous forme cartésienne et polaire. Vérifier en Python que le cube de chaque solution donne bien $-8$.

**Réponse théorique** (à compléter) :

L'équation \(z^3=-8\) s'écrit sous forme polaire :

$$
-8=8e^{j(\pi+2k\pi)}
$$

Les trois solutions sont :

$$
z_k=2e^{j(\pi+2k\pi)/3},\quad k=0,1,2
$$

**Forme polaire :**

$$
z_1=2e^{j\pi/3},\quad
z_2=2e^{j\pi},\quad
z_3=2e^{j5\pi/3}
$$

**Forme cartésienne :**

$$
z_1=1+j\sqrt3,\quad
z_2=-2,\quad
z_3=1-j\sqrt3
$$

Chaque solution vérifie bien \(z^3=-8\).


```python
# Vérification Python

import numpy as np

z1 = 1 + 1j*np.sqrt(3)
z2 = -2 + 0j
z3 = 1 - 1j*np.sqrt(3)

print(np.round(z1**3, 10))
print(np.round(z2**3, 10))
print(np.round(z3**3, 10))
```

### Exercice 3 – Équation avec exponentielle
Trouver tous les réels $\theta$ tels que $e^{j\theta} + e^{-j\theta} = \sqrt{3}$.

Que valent $\cos\theta$ et $\sin\theta$ ?

**Solution** (à compléter) :


On utilise la formule d'Euler :

$$
e^{j\theta}=\cos\theta+j\sin\theta
$$

$$
e^{-j\theta}=\cos\theta-j\sin\theta
$$

En additionnant les deux expressions :

$$
e^{j\theta}+e^{-j\theta}=2\cos\theta
$$

Or, d'après l'énoncé :

$$
2\cos\theta=\sqrt{3}
$$

Donc :

$$
\boxed{\cos\theta=\frac{\sqrt{3}}{2}}
$$

Les solutions réelles sont :

$$
\boxed{\theta=2k\pi\pm\frac{\pi}{6},\quad k\in\mathbb{Z}}
$$

Ainsi :

- Pour $\theta=2k\pi+\frac{\pi}{6}$, $\sin\theta=\frac{1}{2}$.
- Pour $\theta=2k\pi-\frac{\pi}{6}$, $\sin\theta=-\frac{1}{2}$.

avec $k\in\mathbb{Z}$.

Soit $x(t) = 3\cos(2\pi t) + 4\sin(2\pi t)$.

$$
x(t)=3\cos(2\pi t)+4\sin(2\pi t)

x(t)=A\cos(2\pi t-\phi)

### Exercice 6 – Battements
Deux signaux de fréquences proches :  
$
x_1(t) = \cos(2\pi f_1 t)
$ et 
$
x_2(t) = \cos(2\pi f_2 t)
$.  
Montrer que leur somme peut s’écrire  
$$ 
2\cos(2\pi f_m t)\cos(2\pi f_d t) 
$$  
avec $f_m = \frac{f_1+f_2}{2}$ et $f_d = \frac{f_1-f_2}{2}$.  
Pour $f_1 = 10$ Hz et $f_2 = 12$ Hz, tracer la somme et observer l’enveloppe. Expliquer le phénomène de battement.

**Démonstration** (à compléter) :


### 1. Démonstration

On a :

$$
x(t)=x_1(t)+x_2(t)
$$

$$
x(t)=\cos(2\pi f_1t)+\cos(2\pi f_2t)
$$

En utilisant l'identité trigonométrique :

$$
\cos(a)+\cos(b)
=2\cos\left(\frac{a+b}{2}\right)
\cos\left(\frac{a-b}{2}\right)
$$

On obtient :

$$
x(t)=2\cos\left(\frac{2\pi f_1t+2\pi f_2t}{2}\right)
\cos\left(\frac{2\pi f_1t-2\pi f_2t}{2}\right)
$$

Après simplification :

$$
x(t)=2\cos\left(2\pi\frac{f_1+f_2}{2}t\right)
\cos\left(2\pi\frac{f_1-f_2}{2}t\right)
$$

En posant :

$$
f_m=\frac{f_1+f_2}{2}
$$

$$
f_d=\frac{f_1-f_2}{2}
$$

On obtient finalement :

$$
\boxed{x(t)=2\cos(2\pi f_mt)\cos(2\pi f_dt)}
$$



### 2. Application numérique

Pour \(f_1=10\text{ Hz}\) et \(f_2=12\text{ Hz}\) :

**Calcul de la fréquence moyenne :**

$$
f_m=\frac{10+12}{2}=11\text{ Hz}
$$

**Calcul de la demi-différence des fréquences :**

$$
f_d=\frac{10-12}{2}=-1\text{ Hz}
$$

Comme le cosinus est une fonction paire :

$$
\cos(-2\pi t)=\cos(2\pi t)
$$

La somme s'écrit donc :

$$
\boxed{x(t)=2\cos(2\pi 11t)\cos(2\pi t)}
$$


```python

import numpy as np
import matplotlib.pyplot as plt

f1 = 10
f2 = 12
t = np.linspace(0, 1, 2000)

x1 = np.cos(2 * np.pi * f1 * t)
x2 = np.cos(2 * np.pi * f2 * t)
x_sum = x1 + x2

fm = (f1 + f2) / 2
fd = (f1 - f2) / 2

x_env = 2 * np.abs(np.cos(2 * np.pi * fd * t))

plt.plot(t, x_sum, label="Somme")
plt.plot(t, x_env, "--", label="Enveloppe supérieure")
plt.plot(t, -x_env, "--", label="Enveloppe inférieure")

plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.title("Phénomène de battements")
plt.legend()
plt.grid()
plt.show()

# Explication : la somme de deux sinusoïdes de fréquences proches donne une oscillation rapide (fm) modulée par une enveloppe lente (fd).
```


**Observation :**

La somme de deux sinusoïdes de fréquences proches présente une amplitude variable dans le temps. Ce phénomène est appelé battement.

La fréquence des battements est :

$$
f_b=|f_1-f_2|=|12-10|=2\text{ Hz}
$$

La période des battements est :

$$
T_b=\frac{1}{f_b}=0.5\text{ s}
$$

L'enveloppe varie périodiquement et délimite l'amplitude du signal résultant.

### Exercice 8 – Modulation d’amplitude
Un signal modulé en amplitude s’écrit  
$$ 
s(t) = A_c \big[1 + m \cos(2\pi f_m t)\big] \cos(2\pi f_c t).
$$  
Générer ce signal pour $f_c = 100$ Hz, $f_m = 5$ Hz, $m = 0.5$, $A_c = 1$.  
Tracer le signal sur 0,2 s, puis calculer sa transformée de Fourier (FFT) pour visualiser les raies spectrales (porteuse et bandes latérales).  
Expliquer pourquoi il y a trois composantes fréquentielles.

**Code** :

```python

import numpy as np
import matplotlib.pyplot as plt

fs = 1000
t = np.linspace(0, 0.2, int(0.2 * fs), endpoint=False)

fc = 100
fm = 5
m = 0.5
Ac = 1

s = Ac * (1 + m * np.cos(2 * np.pi * fm * t)) * np.cos(2 * np.pi * fc * t)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(t, s)
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.title("Signal modulé en amplitude")
plt.grid(True)

N = len(s)
freqs = np.fft.rfftfreq(N, 1 / fs)
spectre = np.abs(np.fft.rfft(s)) / N
spectre[1:-1] *= 2

plt.subplot(1, 2, 2)
plt.stem(freqs, spectre)
plt.xlim(0, 150)
plt.xlabel("Fréquence (Hz)")
plt.ylabel("Amplitude")
plt.title("Spectre du signal modulé")
plt.grid(True)

plt.tight_layout()
plt.show()
```


### Explication : présence de trois composantes fréquentielles

Le signal modulé est :

$$
s(t)=A_c[1+m\cos(2\pi f_m t)]\cos(2\pi f_c t)
$$

En développant :

$$
s(t)=A_c\cos(2\pi f_c t)
+A_cm\cos(2\pi f_m t)\cos(2\pi f_c t)
$$

En utilisant l'identité :

$$
\cos(a)\cos(b)=\frac{1}{2}[\cos(a+b)+\cos(a-b)]
$$

On obtient :

$$
s(t)=A_c\cos(2\pi f_c t)
+\frac{A_cm}{2}\cos(2\pi(f_c+f_m)t)
+\frac{A_cm}{2}\cos(2\pi(f_c-f_m)t)
$$

Avec \(f_c=100\text{ Hz}\), \(f_m=5\text{ Hz}\), \(A_c=1\) et \(m=0.5\) :

**Porteuse :**

$$
f_c=100\text{ Hz},\qquad A_c=1
$$

**Bande latérale supérieure :**

$$
f_c+f_m=100+5=105\text{ Hz}
$$

$$
A_{\text{sup}}=\frac{A_cm}{2}=0.25
$$

**Bande latérale inférieure :**

$$
f_c-f_m=100-5=95\text{ Hz}
$$

$$
A_{\text{inf}}=\frac{A_cm}{2}=0.25
$$

### Conclusion

Le spectre contient trois composantes fréquentielles :

- La porteuse à \(100\text{ Hz}\).
- La bande latérale inférieure à \(95\text{ Hz}\).
- La bande latérale supérieure à \(105\text{ Hz}\).

Ces trois composantes apparaissent à cause du produit entre la porteuse et le signal modulant.


### Exercice 9 – Repliement de spectre (aliasing) – approche numérique
Échantillonner le signal $x(t) = \cos(2\pi 30 t)$ avec une fréquence d’échantillonnage $f_e = 20$ Hz (donc sous-échantillonnage).  
Générer les échantillons pour $t = 0, 0.05, 0.10, \dots$ et les tracer.  
Reconstruire le signal à partir des échantillons en traçant la droite reliant les points (ou en utilisant une interpolation).  
Quelle fréquence apparente observez-vous ? Justifier à l’aide de la formule d’aliasing.

**Code** :

Repliement de spectre (aliasing)

Le signal original est :

$$
x(t)=\cos(2\pi f t)
$$

avec :

$$
f=30\text{ Hz}, \qquad f_e=20\text{ Hz}
$$

La période d'échantillonnage est :

$$
T_e=\frac{1}{f_e}=\frac{1}{20}=0.05\text{ s}
$$

Les instants d'échantillonnage sont :

$$
t_n=nT_e
$$

Le signal échantillonné est donc :

$$
x[n]=\cos\left(2\pi\frac{f}{f_e}n\right)
$$

En remplaçant les valeurs :

$$
x[n]=\cos\left(2\pi\frac{30}{20}n\right)
$$

$$
x[n]=\cos(3\pi n)
$$

Or, pour tout entier n :

$$
\cos(3\pi n)=\cos(\pi n)
$$

Donc :

$$
x[n]=\cos(\pi n)=(-1)^n
$$

### Fréquence apparente (aliasing)

La formule de la fréquence alias est :

$$
f_{\text{alias}}=|f-kf_e|
$$

On choisit l'entier \(k=1\) :

$$
f_{\text{alias}}=|30-20|=10\text{ Hz}
$$

La fréquence apparente observée est donc :

$$
\boxed{f_{\text{alias}}=10\text{ Hz}}
$$

### Conclusion

Le signal original possède une fréquence de 30 Hz, mais après l'échantillonnage à 20 Hz, il apparaît comme un signal de fréquence 10 Hz.

Cela est dû au sous-échantillonnage : la fréquence d'échantillonnage est inférieure au double de la fréquence du signal, donc le phénomène d'aliasing apparaît.

```python
import numpy as np
import matplotlib.pyplot as plt

fe = 20
f = 30

# Signal original
t_cont = np.linspace(0, 0.5, 1000)
x_cont = np.cos(2 * np.pi * f * t_cont)

# Échantillonnage
te = 1 / fe
t_ech = np.arange(0, 0.5, te)
x_ech = np.cos(2 * np.pi * f * t_ech)

# Reconstruction par interpolation linéaire
t_recon = np.linspace(t_ech[0], t_ech[-1], 1000)
x_recon = np.interp(t_recon, t_ech, x_ech)

# Tracé
plt.figure(figsize=(12, 5))

plt.plot(t_cont, x_cont, label="Signal original (30 Hz)")
plt.stem(
    t_ech,
    x_ech,
    linefmt="r-",
    markerfmt="ro",
    basefmt=" ",
    label="Échantillons"
)
plt.plot(t_recon, x_recon, "--", label="Reconstruction linéaire")

plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.title("Repliement de spectre (aliasing)")
plt.legend()
plt.grid(True)
plt.show()
```

### Exercice 10 – Synthèse d’un signal carré par séries de Fourier
Un signal carré de période $T$ et d’amplitude 1 peut être décomposé en série de Fourier :  
$$
x(t) = \frac{4}{\pi} \sum_{k=0}^{N} \frac{1}{2k+1} \sin\bigl(2\pi (2k+1) f_0 t\bigr).
$$  
Écrire une fonction Python qui génère ce signal pour un nombre d’harmoniques $N$ donné.  
Tester pour $N=1, 5, 20, 50$ et tracer les signaux correspondants.  
Observer le phénomène de Gibbs (oscillations près des discontinuités).  
Comparer avec la fonction `square` de `scipy.signal`.

**Code** :

```python

import numpy as np
import matplotlib.pyplot as plt

def signal_carre_fourier(t, f0, N):
    x = np.zeros_like(t)

    for k in range(N + 1):
        x += (1 / (2 * k + 1)) * np.sin(
            2 * np.pi * (2 * k + 1) * f0 * t
        )

    return (4 / np.pi) * x


T = 1
f0 = 1 / T
t = np.linspace(0, 2 * T, 2000)

# Signal carré idéal sans scipy
x_carre = np.sign(np.sin(2 * np.pi * f0 * t))

# Tracé pour N = 1, 5, 20 et 50
valeurs_N = [1, 5, 20, 50]

plt.figure(figsize=(12, 10))

for i, N in enumerate(valeurs_N):
    x = signal_carre_fourier(t, f0, N)

    plt.subplot(4, 1, i + 1)
    plt.plot(t, x, label=f"Série de Fourier, N={N}")
    plt.plot(t, x_carre, "r--", label="Signal carré idéal")
    plt.ylabel("Amplitude")
    plt.grid(True)
    plt.legend()

plt.xlabel("Temps (s)")
plt.tight_layout()
plt.show()
```

### Exercice 11 – Phase et décalage temporel
Montrer que $A\cos(2\pi f t - \phi)$ est équivalent à un décalage temporel. Quel est le décalage en secondes ?  
Pour $A=1, f=2$ Hz, $\phi = \pi/2$, calculer le décalage. Tracer le signal non décalé et le signal décalé pour vérifier.

**Démonstration** : $\cos(2\pi f (t - \Delta t)) = \cos(2\pi f t - 2\pi f \Delta t)$ donc $\phi = 2\pi f \Delta t$ et $\Delta t = \phi/(2\pi f)$.

**Code** :

Phase et décalage temporel


Le signal décalé est défini par :

$$
x_d(t)=A\cos(2\pi ft-\phi)
$$

On peut écrire :

$$
x_d(t)=A\cos\left(2\pi f\left(t-\frac{\phi}{2\pi f}\right)\right)
$$

En comparant avec un signal retardé :

$$
x_d(t)=A\cos(2\pi f(t-\Delta t))
$$

On obtient :

$$
\phi=2\pi f\Delta t
$$

Donc, le décalage temporel est :

$$
\Delta t=\frac{\phi}{2\pi f}
$$

### Application numérique

Les données sont :

$$
A=1,\qquad f=2\text{ Hz},\qquad \phi=\frac{\pi}{2}
$$

Le décalage temporel vaut :

$$
\Delta t=\frac{\pi/2}{2\pi\times 2}
$$

$$
\Delta t=\frac{1}{8}=0.125\text{ s}
$$

### Conclusion

Le signal décalé est retardé de 0.125 s, soit 125 ms, par rapport au signal original.

Une phase positive dans l'expression \(A\cos(2\pi ft-\phi)\) correspond à un retard temporel.


```python

import numpy as np
import matplotlib.pyplot as plt

A = 1
f = 2
phi = np.pi / 2

dt = phi / (2 * np.pi * f)

t = np.linspace(0, 1.5, 1000)

x_original = A * np.cos(2 * np.pi * f * t)
x_decale = A * np.cos(2 * np.pi * f * t - phi)

plt.figure(figsize=(10, 5))

plt.plot(t, x_original, label="Signal non décalé")
plt.plot(t, x_decale, "--", label="Signal décalé")

plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.title(f"Décalage temporel : Δt = {dt:.3f} s")
plt.legend()
plt.grid(True)
plt.show()

print("Décalage temporel =", dt, "s")
print("Décalage temporel =", dt * 1000, "ms")
```

### Exercice 12 – Échantillonnage d’une exponentielle complexe
Soit $z[n] = e^{j\omega_0 n}$. Montrer que si $\omega_0$ est un multiple rationnel de $2\pi$, le signal est périodique.  
Pour $\omega_0 = 2\pi/3$, déterminer la période fondamentale.  
Pour $\omega_0 = 1$ rad/échantillon, le signal est-il périodique ? Justifier.  
Générer les deux signaux sur 50 échantillons et observer leur comportement.

**Démonstration** : La périodicité exige $\omega_0 N = 2\pi k$ avec $k$ entier, donc $\omega_0/(2\pi) = k/N$ est rationnel.

**Code** :


###   Échantillonnage d’une exponentielle complexe

On considère le signal :

$$
z[n]=e^{j\omega_0 n}
$$

Un signal discret est périodique s'il existe un entier positif \(N\) tel que :

$$
z[n+N]=z[n]
$$

Donc :

$$
e^{j\omega_0(n+N)}=e^{j\omega_0 n}
$$

En simplifiant :

$$
e^{j\omega_0 N}=1
$$

Cela exige :

$$
\omega_0 N=2\pi k
$$

avec \(k\) entier. Ainsi :

$$
\frac{\omega_0}{2\pi}=\frac{k}{N}
$$

Le signal est périodique si et seulement si \(\frac{\omega_0}{2\pi}\) est rationnel.

### Cas 1 : \(\omega_0=\frac{2\pi}{3}\)

On cherche la période fondamentale \(N_0\) :

$$
\frac{2\pi}{3}N_0=2\pi k
$$

$$
N_0=3k
$$

La plus petite valeur entière positive est obtenue pour \(k=1\) :

$$
\boxed{N_0=3}
$$

Le signal est donc périodique de période fondamentale 3 échantillons.

### Cas 2 : \(\omega_0=1\) rad/échantillon

On a :

$$
\frac{\omega_0}{2\pi}=\frac{1}{2\pi}
$$

Comme \(\frac{1}{2\pi}\) est irrationnel, aucun entier positif \(N\) ne peut satisfaire la condition de périodicité.

$$
\boxed{\text{Le signal n'est pas périodique}}
$$

### Conclusion

- Pour \(\omega_0=\frac{2\pi}{3}\), le signal est périodique de période fondamentale \(N_0=3\).
- Pour \(\omega_0=1\), le signal est apériodique.


```python

n = np.arange(0, 50)

omega1 = 2*np.pi/3
z1 = np.exp(1j * omega1 * n)

omega2 = 1
z2 = np.exp(1j * omega2 * n)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.stem(n, np.real(z1), basefmt=' ')
plt.title(r'$\omega_0 = 2\pi/3$ (périodique, période 3)')

plt.subplot(1, 2, 2)
plt.stem(n, np.real(z2), basefmt=' ')
plt.title(r'$\omega_0 = 1$ (non périodique)')

plt.tight_layout()
plt.show()

# Vérification : pour omega1, la période est 3 car 2pi/3 * 3 = 2pi.
```
